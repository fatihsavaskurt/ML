```python
y_train_binary = y_train.apply(lambda x: 1 if x.strip() == '>50K' else 0)
```

```python
y_train_binary.unique()
```

```python
target_means = y_train_binary.groupby(X_train["native-country"]).mean()
```

```python
X_train["native-county-encoded"] = X_train["native-country"].map(target_means)
X_train["native-county-encoded"] = X_train["native-county-encoded"].fillna(y_train_binary.mean())

X_test["native-county-encoded"] = X_test["native-country"].map(target_means)
X_test["native-county-encoded"] = X_test["native-county-encoded"].fillna(y_train_binary.mean())
```

```python
X_train = X_train.drop("native-country", axis= 1)
X_test = X_test.drop("native-country", axis = 1)
```

```python
categorical
```

```python
one_hot_categories = [
    'workclass',
 'education',
 'marital-status',
 'occupation',
 'relationship',
 'race',
 'sex',
]
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
encoder = ColumnTransformer(
    transformers= [
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output= False), one_hot_categories),
    ], remainder= "passthrough"
)
```

```python
X_train_enc = encoder.fit_transform(X_train)
X_test_enc = encoder.transform(X_test)
```

```python
X_train_enc
```

```python
columns = encoder.get_feature_names_out()
```

```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc = RandomForestClassifier(n_estimators= 10, random_state= 15)
rfc.fit(X_train, y_train)
y_pred = rfc.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
print(f'Model accuracy score with default decision-trees : {accuracy_score(y_test, y_pred)}')
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
## Hyperparameter Tuning
rf_params = {"max_depth": [5, 8, 15, None, 10],
             "max_features": [5, 7, "auto", 8],
             "min_samples_split": [2, 8, 15, 20],
             "n_estimators": [100, 200, 500, 1000]}
```

```python
from sklearn.model_selection import RandomizedSearchCV
rfc = RandomForestClassifier()
rscv = RandomizedSearchCV(estimator=rfc, param_distributions=rf_params,n_iter=10,cv=3,verbose=2,n_jobs=-1)
rscv.fit(X_train, y_train)
```

```python
y_pred = rscv.predict(X_test)
print(f'Model accuracy score with default decision-trees : {accuracy_score(y_test, y_pred)}')
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```
