### Encode

```python
df["AMT_INCOME_TOTAL"].describe().T
```

```python
df["DECILE"] = pd.qcut(df["AMT_INCOME_TOTAL"], q = 10, labels = False, duplicates= "drop") +1
```

```python
df.groupby("DECILE").agg(
    COUNT = ("TARGET", "count"),
    TARGET_RATE = ("TARGET", "mean"),
    INCOME_MEAN = ("AMT_INCOME_TOTAL", "mean")
)
```

#### Feature Engineering

```python
df["INCOME/CREDIT"] = df["AMT_INCOME_TOTAL"] / df["AMT_CREDIT"]
```

```python
df["ANNUITY/INCOME"] = df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]
```

```python
df["CREDIT/ANNUITY"] = df["AMT_CREDIT"] / df["AMT_ANNUITY"]
```

```python
df["DECILE_INC_CR"] = pd.qcut(df["INCOME/CREDIT"], q = 10, labels = False, duplicates= "drop") +1
```

```python
df["DECILE_ANN_INC"] = pd.qcut(df["ANNUITY/INCOME"], q = 10, labels = False, duplicates= "drop") +1
```

```python
df["DECILE_CR_ANN"] = pd.qcut(df["CREDIT/ANNUITY"], q = 10, labels = False, duplicates= "drop") +1
```

```python
df.groupby("DECILE_INC_CR").agg(
    TARGET_RATE = ("TARGET", "mean"),
)
```

```python
df.groupby("DECILE_ANN_INC").agg(
    TARGET_RATE = ("TARGET", "mean"),
)
```

```python
df.groupby("DECILE_CR_ANN").agg(
    TARGET_RATE = ("TARGET", "mean"),
)
```

```python
df.groupby("DECILE_CR_ANN").agg(
    COUNT = ("TARGET", "count"),
    MIN = ("CREDIT/ANNUITY", "min"),
    MAX = ("CREDIT/ANNUITY", "max"),
    MEAN = ("CREDIT/ANNUITY", "mean"),
    TARGET_RATE = ("TARGET", "mean")
    
)
```
