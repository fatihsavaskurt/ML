```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/selva86/datasets/master/Advertising.csv"
df = pd.read_csv(url)
```

```python
df
```

```python
X = df.drop(["sales", "Unnamed: 0"], axis = 1)
```

```python
y = df["sales"]
```

```python
X
```

```python
y
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X, y)
```

```python
model.coef_ #b1, b2, b3 değerleri
```

```python
model.intercept_ #b0 değeri
```

```python
model.predict([[460,85,65],[200,45,23]])
```

```python
y_pred = model.predict(X)
```

```python
list(zip(y, y_pred))
```

```python
np.mean((y-y_pred)**2)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mean_squared_error(y, y_pred)
```

```python
np.mean((y-np.mean(y))**2)
```
